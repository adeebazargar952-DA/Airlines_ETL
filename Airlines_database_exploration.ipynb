```python
profiles = []

for table in tables["name"]:

    df = pd.read_sql(
        f"SELECT * FROM {table}",
        conn
    )

    profile = {
        "Table": table,
        "Rows": df.shape[0],
        "Columns": df.shape[1],
        "Missing Values": df.isnull().sum().sum(),
        "Duplicate Rows": df.duplicated().sum(),
        "Memory (MB)": round(
            df.memory_usage(deep=True).sum() / 1024**2,
            2
        )
    }

    profiles.append(profile)

profile_df = pd.DataFrame(profiles)

profile_df = profile_df.sort_values(
    by="Rows",
    ascending=False
)

profile_df = profile_df.reset_index(drop=True)

profile_df
```

```python
# Q1: Which routes generate the highest total revenue?
# Route is treated as a bidirectional airport pair.


route_revenue = pd.read_sql_query("""
    SELECT
        CASE
            WHEN f.departure_airport < f.arrival_airport
            THEN f.departure_airport || ' ↔ ' || f.arrival_airport
            ELSE f.arrival_airport || ' ↔ ' || f.departure_airport
        END AS route,

        SUM(tf.amount) AS total_revenue

    FROM ticket_flights AS tf

    JOIN flights AS f
        ON tf.flight_id = f.flight_id

    GROUP BY
        CASE
            WHEN f.departure_airport < f.arrival_airport
            THEN f.departure_airport || ' ↔ ' || f.arrival_airport
            ELSE f.arrival_airport || ' ↔ ' || f.departure_airport
        END

    ORDER BY total_revenue DESC
""", conn)

route_revenue.head(10)


route_revenue.shape


route_revenue.info()


route_revenue.head(10)
```

```python
Q:What proportion of total revenue comes from each route, and how quickly does cumulative revenue build up?


# Calculate total revenue across all routes
total_revenue = route_revenue["total_revenue"].sum()

# Calculate each route's percentage contribution to total revenue
route_revenue["revenue_share"] = (
    route_revenue["total_revenue"] / total_revenue * 100
)

# Calculate cumulative revenue contribution as we move down the ranked routes
route_revenue["cumulative_revenue_share"] = (
    route_revenue["revenue_share"].cumsum()
)

# Display the top 10 routes
route_revenue.head(10)
```
